```python
N=int(input("Give the upper limit of checking the Collatz conjecture: "))
for i in range(1,N+1):
  print(i)
  k=i
  while k!=1:
    if k%2==0:
      k=k/2
      print(k)
    else:
      k=3*k+1
      print(k)
  print("The conjecture is true for: ",i)
```
